# Extract ACLF Samples

The ACLF condition significantly alters the genetic makeup of cells, which in turn affects their expression. This can lead to extensive cell death, particularly in liver-specific cells. However, endothelial cells may not be as affected by this process, as they continue to receive support and may not be involved in the signaling pathways that trigger cell death in hepatic stellate cells.

## Load Libraries

```python
# Standard imports
import datetime
import os

# Third-party imports
import anndata as ad
```

## Load Parameters

```python
data_dir = input("Enter data directory: ")

# Implement fall back
if data_dir == "":
    data_dir = "../data/h5ad_objects/"

print(f"💾 Data extracted from: {data_dir}")
```

```python
# AnnData file name
data_file = input("Enter file name:")

print(f"💾 Data stored in: {data_file}")
```

## Load Data

```python
adata = ad.read_h5ad(os.path.join(data_dir, data_file))
adata
```

## ACLF Extraction

Extract all samples with disease condition ACLF from the dataset. These samples may introduce problematic signals.

```python
# Store total samples for later calculations
total_samples = adata.n_obs
total_samples
```

```python
# Create subset of non-ACLF samples
non_aclf_adata = adata[adata.obs["condition"] != "aclf"]

# Calculate number of non-ACLF samples
non_aclf_samples = non_aclf_adata.n_obs
non_aclf_samples
```

```python
extracted_samples = ((total_samples - non_aclf_samples) / total_samples) * 100

print(f"A total of {extracted_samples:.2f}% samples is removed.")
```
