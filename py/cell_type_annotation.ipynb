# Cell Type Annotation

Performs cell type annotation on the merged liver `AnnData object`.

## Load Libraries

```python
# Standard imports
import os

# Third-party imports
import anndata as ad
import scanpy as sc
```

## Load Parameters

```python
data_dir = input("Enter data directory: ")

# Implement fall back
if data_dir == "":
    data_dir = "../data/h5ad_objects/"

print(f"💾 Data extracted from: {data_dir}")
```

```python
# AnnData file name
data_file = input("Enter file name:")

print(f"💾 Data stored in: {data_file}")
```

## Annotation With `CellTypist`

The online interface for `CellTypist` requires a **h5ad** file with a log-normalized expression matrix (to 10,000 counts per cell). Process this with `scanpy`.

### Load Data

```python
adata = ad.read_h5ad(os.path.join(data_dir, data_file))
adata
```

### Normalize Data

```python
# Store raw counts
adata.raw = adata
```

```python
adata.raw.X[:5, :5].todense()
```

```python
# Normalize counts to 10,000 per cell
sc.pp.normalize_total(adata, target_sum=1e4)

# Log-transform the normalized data (log(1 + X))
sc.pp.log1p(adata)
```

```python
adata.X[:5, :5].todense()
```

### Save Object

```python
adata
```

```python
adata.write(os.path.join(data_dir, data_file))
```
